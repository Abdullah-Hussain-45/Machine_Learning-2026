```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

df = pd.read_csv("Cleaned train.csv")
# If dataset contains any Id remove it
if "Id" in df.columns:
    df = df.drop(columns=["Id"])
    
df.shape

# Separate Target and features from dataset
X = df.drop(columns=["Saleprice"]) # X as input feature
y = df["Saleprice"] # y as target feature(output)

plt.figure(figsize=(8,4))
sns.histplot(y, kde=True, color="skyblue")
plt.title("Total Distribution: Sale Price")
plt.xlabel("Sale Price($)")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

print(f"Training Size: {X_train.shape[0]} rows")
print(f"Testing Size: {X_test.shape[0]} rows")
```
